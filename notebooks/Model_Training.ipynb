```python
from ultralytics import YOLO

# Modeli seçiyoruz (Small model)
model = YOLO('yolov8s.pt')

# Eğitimi başlatıyoruz (Dosya yolları Oruntu_Proje'ye göre ayarlandı)
results = model.train(
    data='/content/dataset/dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    project='/content/drive/MyDrive/Oruntu_Proje',
    name='Egitim_Sonuclari'
)
```

```python
from ultralytics import YOLO

# 1. Beynimizi çağırıyoruz
model = YOLO('/content/drive/MyDrive/Oruntu_Proje/Egitim_Sonuclari-2/weights/best.pt')

# 2. MOV videomuzu işliyoruz
results = model.predict(
    source='/content/drive/MyDrive/Oruntu_Proje/test.mov', # Uzantıyı .mov yaptık! (İsmi kendine göre düzelt)
    save=True,
    conf=0.40,  # Ekran parlaması olacağı için hassasiyeti biraz esnettik (%40 yaptık)
    project='/content/drive/MyDrive/Oruntu_Proje',
    name='Test_Ciktisi'
)
```
